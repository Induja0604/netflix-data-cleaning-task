```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv('Netflix_Raw_Dataset.csv')
df
```

```python
#Removed duplicates
```

```python
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
#drop duplicates
```

```python
df = df.drop_duplicates()
```

```python
#Standardize text values like gender, country names, etc.
```

```python
text_cols = ['type', 'title', 'director', 'cast', 
             'country', 'rating', 'listed_in']

for col in text_cols:
    df[col] = df[col].str.lower().str.strip()
```

```python
df
```

```python
#Convert date formats to a consistent type (e.g., dd-mm-yyyy)
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

df['date_added_formatted'] = df['date_added'].dt.strftime('%d-%m-%Y')
```

```python
df
```

```python
#Rename column headers to be clean and uniform (e.g., lowercase, no spaces)
```

```python
df.columns = df.columns.str.lower()
```
